## Study area and data provenance (v2: catchment-extended)

## Fire as a barrier to land restoration (new in v2)

Rewilding Portugal's 2025 annual review (`RW_01_AnnualReview2025_ENGLISH_FINALWEB.pdf`) documents wildfires burning roughly 10% of the 318,000ha Greater Côa Valley in 2025, hitting the Ermo das Águias rewilding area hardest (a quarter of its area affected, per the review's own "Iberian Peninsula on fire" section), and names "Fire Risk" as one of the project's own monitoring indicators. That's the motivation for this section — but the review itself has no fire-perimeter geometry, only narrative description, so it's cited here as context, not as the data source.

**Data source, and a real dead end worth stating rather than hiding**: EFFIS (European Forest Fire Information System, JRC) was the first choice for real historical burnt-area perimeters. Two things blocked it, checked directly this session: its actual historical GIS archive requires a manual data-request form (`https://effis.jrc.ec.europa.eu/apps/data.request.form`), not an open API; and its live WFS burnt-area layer (`ercc.ba`) returned a server-side error (`msOracleSpatialLayerOpen(): ... Connection failure`) when queried. Confirmed with Linda: used **NASA's MODIS Burned Area Monthly product (MCD64A1.061)** instead — real satellite-derived burn history, served as public, no-auth Cloud-Optimized GeoTIFFs via Microsoft Planetary Computer's STAC API. Same spirit (real observed fire history, not a synthetic flammability proxy), different real source.

**A second data-currency limitation, also worth stating plainly**: Planetary Computer's ingestion of this MODIS product currently tops out at day 182 of 2025 (~1 July 2025) — checked globally, not just for this tile. That means it **cannot** capture the annual review's own 2025 summer fires (if they landed after that date, which the review's "this summer" framing suggests) or the field-note-documented wildfire near Almeida/Sabugal/Pinhel from 27–29 July 2026 (`field-trips/2026-08-16-Wildfire-Damage-Amoreira-Castelo-Mendo.md`). Both stay narrative-only here — the fire layer below is real data, just not current through today.

**What the real data shows** (2015 – 1 July 2025, 124 monthly MODIS items covering this tile): **10.07% of the extended study area's grid cells show at least one detected burn** in this ~10-year window. The largest single-month burn events fall in 2015, 2017, and 2022 — all independently known as severe Portugal wildfire seasons nationally, which is a reasonable real-world sanity check that this dataset is picking up genuine signal, not noise.

## Field-observed barriers and heritage protection as resistance penalties (new in v4)

Two new inputs feed the Land and Water resistance surfaces this version, on top of the fire-barrier penalty above (still land-group-only, unchanged).

**Richer field-observation export.** `acquire_field_observations.py` now reads `geoData/CSV.zip` (`survey_0.csv`) instead of the earlier FGDB export — a later, richer pull of the same Survey123 "coa-connectivity-lab-survey" form. The first 12 rows are byte-identical to the old export's 12 `visited_sites` records; the CSV adds 10 more site visits, plus up to three repeated species- and barrier-observation blocks per row (Survey123's repeat-group export convention). It now writes **three** layers to `field_observations.gpkg`: `visited_sites` (22 rows), `species_observations` (26 rows, narrative-only — species identity here is often uncertain in the field notes themselves, e.g. "possible Sorraia phenotype, unconfirmed," and this layer is deliberately *not* fed into `build_suitability_models.py` or reconciled against the GBIF-modelled species list), and `barrier_observations` (25 rows melted from up to three per-visit barrier blocks; 22 are `resistance_eligible` — 2 rows excluded as fire/vegetation-disturbance notes mislabeled under Survey123's "other" barrier type, 1 excluded as an unresolvable "other" row with no free text to fall back on; 2 of those 22 eligible rows fall outside `study_area.gpkg`'s bounding box and won't rasterize onto the resistance grid).

**A new, independently sourced heritage layer.** `acquire_unesco_heritage_zones.py` pulls real authoritative boundaries from Portugal's own heritage authority, DGPC (`geo.patrimoniocultural.gov.pt`'s classified-property REST service) — not an estimate, not a buffer-around-coordinates fallback — after UNESCO's own site pages returned a Cloudflare bot-check and the WDPA/Protected Planet API came back empty (both are *cultural*, not natural/mixed, World Heritage sites — a real but unsurprising dead end given WDPA's own scope, not an alarming one). It writes `heritage_protection.gpkg`: `unesco_alto_douro` (24,600.0 ha — an exact match to the published Alto Douro Wine Region figure) and `coa_rock_art_core` (568.6 ha — the rock-art core protected zone specifically, not DGPC's separate, larger buffer/ZEP layer).

**Two new penalty mechanisms in `build_resistance_surfaces.py`**, applied after the existing fire penalty and both capped at the same 100-point resistance ceiling the fire penalty already respects:
- `barrier_penalty()` — a point-distance-decay penalty from each `resistance_eligible` barrier observation, tight and local (500m decay length, vs. the existing road-access kernel's 5km), permeability-tiered (Fully blocking 70 / Partially crossable 30 / Easily crossable 5 peak weight, worst-nearby-barrier wins across tiers), mapped to the Land and/or Water group depending on barrier type (terrestrial barrier types → Land, Dam/Weir/Culvert → Water, Bridge → both). This also fills a gap `build_tradeoff_maps.py`'s own docstring names — the waterway trade-off previously had no dedicated dam/weir point layer and reused the Water resistance surface wholesale.
- `heritage_conflict_penalty()` — an area-fill penalty, *not* distance-decayed: the two heritage polygons above are rasterized directly onto the Land group's grid (`rasterio.features.rasterize`) and every cell inside gets a flat, near-ceiling 90-point penalty — a legally protected heritage landscape is uniformly poor/unavailable matrix across its whole extent, not something that fades out from an edge the way a fence does.

**Air is deliberately excluded from both mechanisms** — stated as an open item, not a silent omission: raptors aren't meaningfully barred by a fence or a vineyard terrace the way a fox or an otter is, but a heritage zone's disturbance regime (tourism activity, low-altitude aircraft/drone restrictions near rock-art panels) could plausibly matter to Air connectivity too. Not modelled here, left for a future iteration.

## Limpopo, South Africa — strategic lessons for expanding the Greater Côa Valley initiative

Rewilding Portugal's own material repeatedly reaches for South African/East African imagery to describe the Côa Valley (the savanna-like structure noted in this project's own field notes; roe deer "repopulating valleys and clearings" language echoing African rewilding narratives). This section takes that comparison seriously and literally: how do savanna-landscape rewilding, land restoration, and ecotourism initiatives in Limpopo province, South Africa, actually work — legally, financially, and operationally — and what, concretely, could the Côa Valley initiative adapt from them to expand its own reach? Built from real South African government/policy/academic sources (cited by URL throughout). Most of this section is not first-hand Limpopo fieldwork, unlike the Côa Valley sections above — with one specific exception, flagged where it appears below: the land-restoration/water-catchment regulatory material is Linda's own professional territory, not literature review. These are strategic *options* worth Rewilding Portugal evaluating, not decisions this notebook makes on the organisation's behalf.

### Connectivity & rewilding method: fence removal vs. circuit theory

The [Associated Private Nature Reserves](https://en.wikipedia.org/wiki/Associated_Private_Nature_Reserves) (APNR — Balule, Kapama, Klaserie, Timbavati, Thornybush, Umbabat; 1,800km²/180,000ha) and the wider [Great Limpopo Transfrontier Conservation Area](https://www.peaceparks.org/tfcas/great-limpopo/) (GLTFCA — South Africa/Mozambique/Zimbabwe, ~10,000,000ha) scaled up almost entirely through **physical fence removal between neighbouring landowners**, not land purchase: the Kruger/APNR boundary fence came down in June 1993, and at Thornybush alone, dropping fences along the reserve's eastern side in 2017 reconnected more than 140km² with adjacent reserves and Kruger, with the elephant count surging from 55 to 770 between 2017 and 2019 ([NASA Earth Observatory](https://science.nasa.gov/earth/earth-observatory/rewilding-south-africas-greater-kruger)). Worth stating plainly rather than presenting fence removal as an unambiguous win: that same NASA source notes the resulting high elephant density drove substantial woody-vegetation loss (browsing, bark-stripping, uprooting) — a real trade-off of fast connectivity gains, not a free lunch.

This is a fundamentally different (and, per-hectare, far cheaper) scaling mechanism than Rewilding Portugal's current model of direct land purchase — 997ha bought to date toward a 5,000ha-by-2030 target, per the [Endangered Landscapes & Seascapes Programme project page](https://www.endangeredlandscapes.org/project/greater-coa-valley/) (`references/ecological-information/endangeredlandscapes-org-project-greater-coa-valley.pdf`, already in this repo).

**Implication for expansion**: land-tenure coordination and fencing/access agreements with neighbouring private landowners along the traced tributary catchment (map in the Study Area section above) could be a lower-cost connectivity-scaling lever than land acquisition alone — with the Thornybush vegetation-loss caveat above as a reason to pair any such agreement with active grazer/browser-density management, not treat "remove the barrier" as the whole plan. Separately: GLTFCA's cross-border, three-country model is a concrete long-horizon reference for a possible Portugal–Spain transfrontier extension, since the Côa/Douro system already crosses into Spain (Arribes del Duero, named in this notebook's own Water section since v1).

### Land restoration as a legal obligation vs. a grant-funded initiative

South Africa's Mineral and Petroleum Resources Development Act (MPRDA, 2002) originally required mining companies to post financial provision (cash deposits, bank guarantees, or trust funds) against future rehabilitation costs; that requirement has since been repealed and replaced by **[NEMA Section 24P](https://cer.org.za/news/what-you-should-know-about-proposed-new-rules-for-financial-provision-for-mining-rehabilitation)**, jointly administered by the Department of Mineral Resources and Energy (DMRE, the competent implementing authority) and the Department of Forestry, Fisheries and the Environment (DFFE, the legislative authority) — see also [Mining Weekly's summary of the legal requirements for mine closure](https://www.miningweekly.com/article/legal-requirements-for-mine-closure-in-south-africa-2023-07-06). This makes rehabilitation a **legally bonded financial obligation on the party causing the degradation**, not a grant-dependent voluntary undertaking. The concrete Limpopo case: [Palabora Mining Company](https://www.miningweekly.com/article/mining-beyond-profits-shows-commitment-to-communities-environment-and-wildlife-conservation-2023-12-15) (copper, directly adjacent to Kruger National Park) runs tailings backfill, progressive native revegetation, and a Biodiversity Action Plan for its Kruger-adjacent Big Five/wildlife coexistence, under ISO 14001 certification.

**This is Linda's own professional territory, not desk research**: as Environmental Deputy Director, Department of Forestry, Fisheries & the Environment, South Africa (1997–2002), she led post-mining land restoration programmes and integrated water-catchment management/environmental governance at regional scale — the same regulatory terrain described above (mine rehabilitation as a bonded obligation, catchment-scale water governance), one policy generation earlier than the current MPRDA/NEMA §24P framework.

There's also a direct methodological echo worth naming: both this notebook's fire-barrier work above and South Africa's mine-rehabilitation regime build a resistance/obligation model around an **actively degrading force** — wildfire here, extraction there — just a different forcing mechanism and a very different institutional backing (a real MODIS-derived recency penalty on a voluntarily-funded resistance surface, vs. a legally bonded closure obligation).

**Implication for expansion**: a legally-mandated financial-provisioning mechanism — for wildfire-adjacent land degradation, or for any future extractive/development pressure near the Côa Valley — is a genuine EU/Portuguese policy-advocacy angle Rewilding Portugal could pursue for funding resilience beyond Endangered Landscapes-style grant dependency. Stated as a real option worth raising, not as something already planned or in motion.

### Water governance: a binding ecological Reserve vs. a stalled protocol

South Africa's **[National Water Act, 1998 (Act 36 of 1998)](https://faolex.fao.org/docs/pdf/saf123836.pdf)** created "the Reserve" — a legally binding minimum water allocation covering both basic human needs and ecosystem health, which takes precedence over water-use rights and was explicitly designed to correct apartheid-era water inequity favouring large-scale white farmers over small-scale Black farmers. That's a meaningfully stronger legal instrument than anything currently governing the Côa's own water: the EU Water Framework Directive water-bodies layer already in this project's pipeline (`data/processed/hydrology_and_protected_areas.gpkg`) is descriptive, not a binding minimum-flow guarantee, and the protocol to remove the Foz Côa cofferdam — signed by Rewilding Portugal, the Portuguese Environment Agency, and the Côa Parque Foundation back in 2023 — is still "awaiting approval" per the 2025 annual review, three years on.

**Implication for expansion**: the National Water Act's Reserve is a citable international precedent for a binding minimum-ecological-flow allocation — useful, real leverage in advocacy for finally moving the stalled cofferdam-removal protocol forward, not just a passive parallel.


### Rights of nature, deepened: what South Africa's Reserve actually is, and why Portugal has nothing comparable (new in v4)

The Water Governance subsection just above already names South Africa's National Water Act Reserve as "a meaningfully stronger legal instrument than anything currently governing the Côa's own water," and leaves it there, as one line in a summary table. A dedicated desk-research pass, [`field-trips/deskStudy/rights-of-nature-portugal-vs-south-africa.md`](../../../field-trips/deskStudy/rights-of-nature-portugal-vs-south-africa.md), goes further on two questions that line doesn't ask — this subsection extends it rather than repeating it.

**What the Reserve precisely is.** The National Water Act's Chapter 3 Reserve has two components — a basic human-needs reserve and an ecological reserve (water required to protect the aquatic ecosystem's own quantity and quality) — and once the Minister determines it for a given water resource, it is legally binding and takes the *highest* priority in South African water allocation: it cannot itself be allocated away for irrigation, industrial, or mining use, and other uses are allocated only from what remains. It was written to correct a specific, named wrong — under apartheid, the white minority held roughly 87% of the land and the water infrastructure attached to it, and by 1994 "hardly any Black individual had a formal water resource right, not even for basic needs." Two qualifications matter, in keeping with this project's own habit of not overclaiming: this is **a minimum-allocation guarantee, not legal personhood** — the river or aquifer isn't itself a party who can sue, the state is legally obliged to protect a defined minimum flow on the ecosystem's behalf, which is a meaningfully different, and less far-reaching, legal category from Ecuador's *Pachamama* rights or New Zealand's Te Awa Tupua river-personhood law; and the **law being strong on paper hasn't closed the gap it was built to close** — decades on, "the biggest users of water remain white commercial farmers," per multiple post-apartheid water-governance reviews.

**Portugal and the EU: essentially nothing comparable exists.** No constitutional provision, statute, or case law grants legal personhood or a binding minimum-allocation guarantee to any river or ecosystem in Portugal — it doesn't appear on any of the surveyed "which countries grant rights to nature" lists checked for this note. The EU Water Framework Directive layer already in this project's own pipeline (`data/processed/hydrology_and_protected_areas.gpkg`) remains descriptive, not a binding minimum-flow guarantee. The closest things that exist are a 2017 draft EU "Nature's Rights" directive that was never adopted, and a 2026 European Citizens' Initiative ("Rights for Nature: Empower Citizens to Represent and Protect Ecosystems") now at the signature-collection stage — a live proposal, not law, with no guarantee of adoption. One concrete indicator of how far that legal gap reaches locally: Faia Brava — founded in 2000 by ATN (renamed "FAIA BRAVA – Nature Conservation Association" in 2024), and a genuinely separate organisation from Rewilding Portugal (founded 2019, Rewilding Europe's local partner for the wider Greater Côa Valley initiative, holding its own directly purchased land at Ermo das Águias and Vale Carapito) — is described in multiple sources as **Portugal's only private protected area**. That fact is really a statement about what state or EU-level legal protection *doesn't* reach: even the valley's flagship voluntarily protected reserve exists because a private actor stepped in, not because Portuguese or EU law required a minimum ecological floor.

**Why the Greater Côa Valley has next to no wildlife.** That phrase needs its own qualification: not literally zero, and the direction over the last few years is recovery, not further decline — the European beaver, confirmed present for the first time in "more than 500 years" via natural dispersal from Spain (first confirmed 2025); cinereous vultures holding 15 nests in neighbouring Serra da Malcata; griffon and Egyptian vultures, Bonelli's and golden eagles, and black storks logged at Faia Brava; the Iberian wolf confirmed at Ermo das Águias under the EU LIFE WolfLux project, per an August 2026 field confirmation. But a meaningful share of what's present is either a managed reintroduction (Sorraia horse and Tauros are semi-wild released grazers, not wild recolonizers) or dependent on active cross-border monitoring — the wolf entry itself names an EU LIFE-funded project as the reason it's trackable here at all. The scarcity this recovery is climbing out of has several compounding, well-documented drivers: deforestation followed by CAP-driven abandonment (the valley called "virtually treeless" 60–70 years ago, then hit by some of the highest rural-abandonment rates in Europe as CAP support concentrated in Portugal's coastal lowlands, not its forested interior); sustained predator persecution (systematic wolf poisoning from the early 1800s drove the peninsula-wide population to roughly 400–500 animals by the 1970s, and the wolf now occupies only about 20% of its historical Portuguese range — one of the few areas a population persisted through that collapse sits just south of the Douro, the Côa Valley's own doorstep); hunting as a current, active land use (337 ICNF hunting zones inside the study area, per this notebook's own Land Tenure section, with nothing in Portuguese or EU law subordinating that use to a prioritised ecological minimum); and, tying directly back to the water-governance comparison above, **no legal floor under recovery, only voluntary purchase** — the 997 ha Rewilding Portugal has bought toward its 5,000 ha target, ATN's Faia Brava, the Ermo das Águias horse and Tauros releases, even the cofferdam-removal protocol still "awaiting approval" three years on, have all happened because a private philanthropic actor voluntarily acted, not because any Portuguese or EU law required it. The beaver's return is the one real exception — that happened on its own, by natural cross-border dispersal, independent of any legal mechanism on either side of the border.

**Implication for expansion**: the near-total absence of a Portuguese or EU legal floor — for water, land, or species — is itself a policy-advocacy target worth naming plainly alongside the National Water Act citation already made above, not just a passive contrast to note and move past.

## Wildlife relations, Portugal vs. South Africa: resistance, history, and everyday coexistence (new in v4)

A companion desk-research note, [`field-trips/deskStudy/wildlife-relations-portugal-vs-south-africa.md`](../../../field-trips/deskStudy/wildlife-relations-portugal-vs-south-africa.md), asks how Portuguese and South African publics actually relate to wildlife and conservation, in three registers. The three don't map onto each other symmetrically — two do real comparative work, the third turns out to have no honest Portuguese counterpart at all, and that absence is treated as a finding, not smoothed into a false equivalence.

**1. Resistance to conservation itself.** In the Côa Valley this runs almost entirely through the wolf: livestock predation rose sharply 2012–2016, compensation is slow (a statutory 60-day deadline routinely stretches to one or two years), and Rewilding Portugal's own "From Fear to Favor" attitude survey (220 respondents south of the Douro) found livestock owners specifically negative-to-neutral, with the general public and hunters neutral to positive — a clear split by direct economic exposure, not a broad hunting-versus-wildlife divide. What wasn't found, despite searching directly for it, is any documented organised opposition to Rewilding Portugal's or Faia Brava's land purchases themselves — the reporting instead describes biologists and rewilding organisations buying land parcel by parcel from families already leaving depopulating farms, not land being taken from people who wanted to keep working it. South Africa's version runs through a physical boundary instead: the 2003 removal of Kruger's fencing to open the Great Limpopo corridor is a live source of conflict two decades on — roughly 60 cattle lost near Giyani in 2024 alone to predators and snares crossing an unfenced boundary, with park authorities legally entitled to seize animals that wander in rather than let owners reclaim them. Genuinely different shapes of resistance — compensation for a predator on land people already farm, versus contested access across a controlled boundary — even though both are, at root, about who absorbs the cost of a large predator's presence.

**2. Resistance from historical dispossession.** South Africa's case here is specific and well documented: the Makuleke community, roughly 1,500 people, was forcibly removed by the apartheid government in 1969 from ~25,000 ha now inside Kruger, and had that land restored to a Community Property Association through a 1998 restitution settlement — ownership, tourism-revenue rights, and a joint management board, under a "contractual park" arrangement. **The Côa Valley has no honest equivalent to this, and forcing one would misrepresent both places.** Nobody was removed by the state to make way for a protected area here. The actual sequence ran the other direction — rural exodus from the 1960s onward, driven by emigration and depopulation with no connection to conservation policy, among the highest rural-abandonment rates recorded anywhere in Europe, land abandoned first for unrelated economic reasons, with rewilding organisations acquiring parcels afterward from willing sellers, decades later. There is no racialised removal apparatus, no forced relocation, and consequently no restitution claim or contractual-park negotiation to point to. If there's a genuine parallel, it's a negative one: Kruger's history is conservation causing displacement that then required a formal justice remedy; the Côa's is displacement happening first, for unrelated reasons, with conservation arriving afterward into land people had already chosen to leave.

**3. General coexistence tolerance.** Hunting is a major, active land use in the Côa Valley — 337 ICNF zones (194 associative, 108 municipal, 35 touristic/private) sitting on the same fences as the rewilding narrative, confirmed directly in this project's own field photographs (a "Zona de Caça Associativa" sign at Largo da Praça, a "Zona de Caça Turística" sign at the Paul de Toirões restored-wetland boundary, both 22 August 2026). Yet the attitude survey complicates the intuitive read: hunters were statistically indistinguishable from the general public in wolf attitudes — neutral to positive — it was livestock owners specifically who carried the hostility, so coexistence tolerance here tracks direct economic exposure far more than a general hunting-versus-wildlife divide. South Africa's coexistence landscape is structured differently again, around who benefits economically from wildlife: trophy hunting contributes on the order of $2.5 billion a year nationally, but community benefit-share from that revenue is disputed (one 2013 estimate put it at ~3% of gross revenue), while reserves able to shift toward photographic-tourism conservation levies instead (Timbavati is the cited example) have moved the balance away from hunting. For communities directly bordering the unfenced Kruger boundary, the lived relationship with wildlife increasingly means loss — cattle, crops — rather than either hunting heritage or tourism income.

## Safari-scale upscaling: what would it take? (new in v4)

A companion desk-research note, [`field-trips/deskStudy/safari-upscaling-coa-valley-limpopo.md`](../../../field-trips/deskStudy/safari-upscaling-coa-valley-limpopo.md), asks the comparison this notebook keeps circling literally: what would it take to move the Côa Valley toward a Limpopo-province-style safari-tourism model? The short answer, stated up front: the Côa Valley isn't close, and the gap isn't a matter of ambition or money alone. Four barriers, two negotiable over decades and two that aren't negotiable at all:

**Barrier one — land-tenure fragmentation.** `land_tenure.gpkg`'s 337 hunting-zone parcels sum to 663,158 ha (mean 1,968 ha) — only 6 individual parcels exceed 10,000 ha (a single wolf pack's territory, per van Maanen 2019), and none individually reach the 100,000 ha "mega-rewilding" contiguous-patch threshold Araújo & Alagador (2024, *Current Biology*) define as necessary for long-term large-carnivore/omnivore survival. Dissolving every touching parcel into contiguous blocks does clear that bar geometrically — 2 of 43 dissolved blocks exceed 100,000 ha, together 56.1% of the study area — but those blocks are stitched together out of dozens of separately licensed ZCA/ZCM/ZCT concessions under different associations, municipalities, and private operators, none of them for sale as a unit, none under common management. The land isn't too small in aggregate; no single actor holds enough of it to run one.

**Barrier two — no Big-Five-equivalent species roster.** Iberia has no living or historically native multi-tonne ecosystem-engineer herbivore (no elephant/rhino/hippo/buffalo equivalent), and the confirmed roster tops out at one genuinely wild, elusive, landscape-scale predator (the Iberian wolf) plus semi-wild grazers standing in for extinct megafauna (Sorraia horse, Tauros) and a strong raptor/scavenger guild — a legitimate rewilding story, not a Big-Five-equivalent tourism product, and no land consolidation changes what species exist to put on it.

**Barrier three — road density and the visitor-infrastructure gap.** This project's own `distance_to_road_m.tif` covariate records a median distance to the nearest mapped road of 27,564 m; only 4.4% of cells fall within 250 m of a road. A low-road-density interior landscape is the opposite of what a multi-lodge safari circuit assumes.

**Barrier four — land that isn't fragmented, it's legally unavailable.** This one isn't negotiable through willing-seller purchase at any timescale. The computed figures below quantify it directly.

```python
import geopandas as gpd
import warnings
warnings.filterwarnings("ignore")

study_area = gpd.read_file(config.DATA_PROCESSED / "study_area.gpkg", layer="study_area")
hunting_zones = gpd.read_file(config.DATA_PROCESSED / "land_tenure.gpkg", layer="hunting_zones")
private_reserves = gpd.read_file(config.DATA_PROCESSED / "land_tenure.gpkg", layer="private_reserves")
alto_douro = gpd.read_file(config.DATA_PROCESSED / "heritage_protection.gpkg", layer="unesco_alto_douro").to_crs(study_area.crs)
rock_art = gpd.read_file(config.DATA_PROCESSED / "heritage_protection.gpkg", layer="coa_rock_art_core").to_crs(study_area.crs)

study_area_ha = study_area.geometry.area.sum() / 10_000
hz_ha = hunting_zones.geometry.area.sum() / 10_000
pr_ha = private_reserves.geometry.area.sum() / 10_000

alto_douro_overlap_ha = gpd.overlay(alto_douro, study_area, how="intersection").geometry.area.sum() / 10_000
rock_art_overlap_ha = gpd.overlay(rock_art, study_area, how="intersection").geometry.area.sum() / 10_000
heritage_exclusion_ha = alto_douro_overlap_ha + rock_art_overlap_ha

GLTFCA_HA = 10_000_000
achievable_ceiling_ha = study_area_ha - heritage_exclusion_ha
published_faia_brava_ha = 856.87

print(f"Study area: {study_area_ha:,.0f} ha ({study_area_ha / 100:,.1f} km2)")
print()
print(f"Hunting zones: {len(hunting_zones)} parcels, {hz_ha:,.0f} ha total, "
      f"mean {hz_ha / len(hunting_zones):,.0f} ha ({hz_ha / study_area_ha * 100:.1f}% of study area)")
print(hunting_zones["tenure_type"].value_counts().to_string())
print()
print(f"private_reserves layer (this project's own OSM-sourced pull): {len(private_reserves)} feature(s), "
      f"{pr_ha:,.1f} ha -- {private_reserves['name'].iloc[0]}")
print(f"  vs. Faia Brava's published figure: {published_faia_brava_ha:,.2f} ha "
      f"-- this layer's polygon is a confirmed DATA-QUALITY UNDERCOUNT, not a corrected re-measurement")
print(f"  (Faia Brava is ATN's reserve, a separate organisation from Rewilding Portugal -- see narration below)")
print()
print(f"UNESCO Alto Douro Wine Region overlap with study area: {alto_douro_overlap_ha:,.2f} ha "
      f"({alto_douro_overlap_ha / 24_600 * 100:.1f}% of the 24,600 ha published polygon -- most of it sits outside this study area)")
print(f"Coa rock art core overlap with study area: {rock_art_overlap_ha:,.2f} ha")
print(f"Total real computed heritage exclusion: {heritage_exclusion_ha:,.2f} ha "
      f"({heritage_exclusion_ha / study_area_ha * 100:.2f}% of study area)")
print()
print(f"Achievable ceiling (study area minus heritage exclusion): {achievable_ceiling_ha:,.0f} ha "
      f"= {achievable_ceiling_ha / GLTFCA_HA * 100:.2f}% of GLTFCA scale ({GLTFCA_HA:,.0f} ha)")
print(f"Currently under active, non-purchasable hunting-zone licence: {hz_ha / study_area_ha * 100:.1f}% of study area")
print(f"Land actually held (this layer's undercounted figure): {pr_ha / GLTFCA_HA * 100:.4f}% of GLTFCA scale")
print(f"Land actually held (published Faia Brava figure): {published_faia_brava_ha / GLTFCA_HA * 100:.4f}% of GLTFCA scale")
```

**What scale is actually achievable.** Against the ~10,000,000 ha Great Limpopo Transfrontier Conservation Area (already cited in v3's Limpopo section), the study area's computed extent above (1,023,036 ha, matching v3's own figure independently) is about 10.2% of GLTFCA scale; subtracting the real computed heritage exclusion leaves roughly 10.16% as the absolute geographic ceiling — the area left over if literally every non-heritage-excluded hectare of the study area were somehow assembled under one rewilding tenure. That ceiling is generous to the point of being unreal: 64.8% of the study area is currently under active hunting-zone licence across 337 separate concessions, none of it on the market. GLTFCA's own scaling mechanism — physical fence removal between neighbouring landowners who already held the land, covered in v3's Limpopo section — isn't available here either: the Côa study area doesn't border a national park at that scale, so there's no equivalent fence to remove. Any upscaling has to run on the same private-purchase, willing-seller model Rewilding Portugal already uses.

The private_reserves layer's Faia Brava figure needs to be read carefully rather than taken at face value: this project's own OSM-sourced pull (214.6 ha) is a data-quality undercount against the published 856.87 ha figure, flagged plainly rather than silently used as if reliable. And a genuine organisational distinction matters for reading either figure: **Faia Brava is ATN's reserve** (Associação Transumância e Natureza, renamed "FAIA BRAVA – Nature Conservation Association" in 2024, founded 2000, Portugal's first private nature reserve, officially classified 2010) — **not Rewilding Portugal's own land.** Rewilding Portugal is a separate organisation, founded 2019, Rewilding Europe's local partner for the wider Greater Côa Valley initiative, holding its own directly purchased land at Ermo das Águias and Vale Carapito. The two collaborate closely (species reintroduction, shared Rewilding Europe affiliation, the Wild Côa Network) but aren't the same legal entity. Under whichever real figure is used, both organisations' combined footprint — Faia Brava's ~857 ha plus Rewilding Portugal's 997 ha purchased toward its own 5,000 ha target — comes to roughly 0.019% of GLTFCA scale: two orders of magnitude below even the geometric ceiling computed above, on a private-purchase, willing-seller process this project's own wildlife-relations note already documents as parcel-by-parcel and generational.

## Two Rivers, Two Stories: the Douro dam cascade and the Foz Côa counterfactual (new in v4)

A companion desk-research note, [`field-trips/deskStudy/douro-vs-coa-tourism-dam-counterfactual.md`](../../../field-trips/deskStudy/douro-vs-coa-tourism-dam-counterfactual.md), corrects a figure that had been circulating in this project's own storyMap draft — a claimed "141 dams and weirs" on the Douro — and uses the correction to pose an honest open question the storyMap material hadn't asked.

**The corrected dam count.** There's no single citable source for "141." There are two real counts at different scales: roughly **15 major dams** on the Douro's own mainstem, from its Spanish headwaters to Porto (six of them — Picote, Bemposta, Carrapatelo, Régua/Bagaúste, Valeira, Pocinho — built in sequence from 1953 to 1982, totaling ~1,680 MW, with Crestuma-Lever adding a further 117 MW in 1985 for ~1,800 MW total); and a basin-wide inventory of **roughly 1,200 dams and weirs** across the whole catchment, Spain and Portugal, every tributary included (Cortes et al. 2019, *Water* — 57 dams, 40 mini-hydro installations, 1,105 medium/small barriers, ~25% assessed obsolete). Fifteen answers "how many dams control the Douro's main channel." Twelve hundred answers "how fragmented is the whole basin." Neither is 141.

**Two tourism economies, built on opposite premises.** The Douro's is built on the infrastructure the region chose to keep — the Alto Douro Wine Region's UNESCO inscription (2001, ~24,600 ha of terraced vineyard) anchors a river-cruise sector made possible precisely by the lock-and-dam system above, grown from ~600,000 passengers in 2014 to 1.38 million in 2024 (up 10.6% on 2023), with the cruise-operator authority now capping new large-vessel approvals. The Côa's runs on the opposite premise — an intact valley rather than a controlled one. The Vale do Côa rock-art sites, inscribed as UNESCO World Heritage two years after the dam's 1996 cancellation, drew 401,387 total visitors across the Archaeological Park's first 30 years of operation (opened August 1996) — an annual average in the low tens of thousands, a scarcity-based heritage-tourism model, not a mass-throughput one. It's a different scale by two orders of magnitude, and that contrast is itself the material for a "Two Rivers, Two Stories" chapter: one economy built on damming the river for controlled access by boat, the other on keeping it wild enough that walking to a rock face still means something.

**The open question this can't resolve.** By the time the Côa dam was cancelled in January 1996, the Douro mainstem alone was already carrying close to 1,800 MW of installed hydroelectric capacity, built over three decades, entirely separate from whatever the Côa project would have added — a real, documented backdrop that makes it reasonable to ask whether cancelling one dam on a comparatively small tributary was politically easier because the region's flagship river a few kilometres away was already generating power at that scale. But no contemporaneous EDP or government statement from 1994–96 framing the cancellation explicitly in those terms turned up in this pass, and the academic literature on the case (Coelho et al., *Ecological Economics*) treats it as a heritage-vs-development values conflict resolved through multi-criteria decision analysis, not as a documented energy-supply-adequacy calculation. Stated honestly: this desk-research pass documents the backdrop and cannot prove causation. Whether the Douro cascade's existence gave the 1996 Council of Ministers political room it wouldn't otherwise have had, or whether the decision would have gone the same way regardless, on the strength of the civic campaign and the engravings' evident scientific importance alone, is left as a genuinely open question — worth carrying into any future storyMap chapter as something the reader is invited to sit with, not a conclusion this notebook has proven.

## Renewable energy: an open policy question, and a concrete proposal (new in v4)

Two companion desk-research notes — [`field-trips/deskStudy/renewable-energy-policy-coa-valley.md`](../../../field-trips/deskStudy/renewable-energy-policy-coa-valley.md) and [`field-trips/deskStudy/integrated-infrastructure-proposal-coa-valley.md`](../../../field-trips/deskStudy/integrated-infrastructure-proposal-coa-valley.md) — take up a real strategic question this notebook hasn't touched before: what does the Côa Valley rewilding movement's own stated position on renewable-energy siting actually commit it to, and is there a version of energy infrastructure the movement could actively propose rather than only oppose?

**What Rewilding Portugal actually said.** Its 18 November 2025 statement opposing the SOPHIA solar project (Fundão–Penamacor, near but outside the Côa watershed proper — 3,500 ha total footprint, cutting the "Raia Norte" and "Raia Sul" ecological corridors connecting Gardunha, Raia, and Malcata) is explicit that this is not a blanket rejection of solar power:

> "Rejecting SOPHIA does not mean rejecting solar energy. It means demanding responsible planning, transparency and ecological justice... large-scale solar power plants, if poorly located, replace living ecosystems with dead surfaces, creating ecological deserts in a country that needs to renaturalise, reconnect and restore itself."

— calling instead for mapping already-artificialised land, tax incentives for panels on public/industrial rooftops, and a "Nature-based Energy Transition programme." Siting-conditional, not anti-renewable.

**The open policy question.** Nothing commits Rewilding Portugal, or the wider movement, to a standing position beyond that one project-specific statement — the real choice is between continuing to fight each SOPHIA-shaped proposal case by case, or generalising the siting-conditional logic already stated into a proactive policy position. There's a live national process this could plug into: Portugal's own transposition of the EU's RED III directive, the "Mapa Verde" Zonas de Aceleração para Energias Renováveis, went to public consultation in June 2026 and **already excludes Natura 2000 and other protected areas from acceleration zones outright** — close to what Rewilding Portugal asked for nine months earlier — though environmental reporting on the consultation describes Portugal as still lacking a clear, robust strategy for which zones actually get designated. Other jurisdictions show what a standing position could look like, and what can go wrong with one: the Netherlands' "zonneladder" (rooftops first, agricultural/natural land last, written into most provincial policy since 2020) is well-articulated but documented as under-enforced, with no binding national instrument behind it; Germany's EEG spatial-control regime for ground-mounted solar (since 2004, ~930,000 ha of pre-screened lower-value land) is the most institutionally mature of the three but the hardest to transplant into Portugal's smaller planning apparatus on any short timescale. This is left as a genuinely open strategic tradeoff — case-by-case opposition wins concrete, immediate fights with a motivated coalition; a standing policy position is slower by construction but stops the same fight from having to be refought project by project.

**A concrete proposal, offered as one constructive answer rather than a further restatement of the question.** Two parts, pitched as one integrated package:

- **Solar access canopies over existing reserve access roads** — applying the same, already-proven agrivoltaic engineering (PV arrays elevated 2–4m on structural supports, IEA-PVPS/DOE documented, adapted here from cropland dual-use to a reserve's own access track rather than an existing road-canopy precedent, which doesn't exist in the literature checked for this note) at three candidate sites: Vale Carapito's VC1 trailhead track (Rewilding Portugal, Vilar Maior), Ermo das Águias's Vale de Madeira → Ribeira de Gaiteiros track (Rewilding Portugal), and Faia Brava's approach road near Cidadelhe — the least well-characterised of the three, flagged as needing a dedicated site visit before it could be proposed with any specificity, **and, because Faia Brava is ATN's reserve and not Rewilding Portugal's own land, any actual proposal there would need separate buy-in from ATN as the reserve's managing organisation, not just Rewilding Portugal.** Illustrative scale only: ~0.3–0.6 MW and €30,000–€80,000/yr per site.
- **Photo-tourism lodges modelled on Timbavati's conservation-levy mechanism** (55–62% of Timbavati's budget from photographic-tourism levies, current rate R660/person/night) — explicitly excluding the trophy-hunting share of Timbavati's real budget mix from the proposal, and scaled honestly down: Côa reserves run at roughly 1–2% of Timbavati's 53,396 ha, implying one small 6–12 room lodge per expanding reserve, not a lodge cluster. Candidate siting draws on this project's own `tourism_sites.gpkg`, steered toward the data's own low-connectivity-zone sites (Pontão Manuel José weir, the Foios biodiversity station, Praia Fluvial de Vale das Éguas) and explicitly away from the high-connectivity ones the connectivity model above flags as ecologically valuable movement corridors (Penascosa, Lageosa, Rapoula do Côa). Illustrative levy revenue: €80,000–€225,000/yr per site.

Both halves apply the same principle SOPHIA's own opposition statement already argues for — build on what's already disturbed, not on what's still wild — one to energy, one to tourism infrastructure. Stated plainly, as the source note does: this is a proposal sketch, not an engineering-ready design or feasibility study, and every figure above is illustrative order-of-magnitude, not a real project number.

## Water, Empire, and Portugal's Environmental Legacy (new in v5)

Everything above looks at the Côa Valley's own water/land/air connectivity. This closing section zooms out: Portugal's broader relationship with water, the ocean currents that physically connect Portugal to three former colonies, and the colonial-era ecological policy those currents carried — read through the corrected Indigenous-Science/safe-and-just framing established near the start of this notebook, not a neutral history.

### Portugal's rivers and seas: water as national identity (new in v5)

This project's water story so far is entirely Côa/Douro basin. Portugal's is national and maritime. The Tejo (Tagus), Iberia's longest river at 1,007 km, has an 80,626 km² basin (24,845 km² in Portugal); mean annual discharge at the Almourol gauging station (2006–2019) was 263 m³/s, rising to 440 m³/s at the estuary, Portugal's second-largest river by that measure — the Douro, already documented elsewhere in this notebook, is larger still at roughly 505 m³/s mean estuary discharge. Portugal's own national hydrological monitoring network, SNIRH, run by APA, is documented as "more oriented for flood evaluation" than drought evaluation — a structural gap in a country that also funds EU water-scarcity-adaptation research for the Tagus basin specifically. Full sourcing in [`field-trips/deskStudy/portugal-water-history-rivers-seas.md`](../../../field-trips/deskStudy/portugal-water-history-rivers-seas.md).

### Land to sea: the ocean currents connecting Portugal to Brazil, Angola, and Mozambique (new in v5)

This closing section's throughline, and a genuine echo of this notebook's own circuit-theory connectivity method at ocean-basin scale. Portuguese ships departing the Tejo and Douro estuaries above used the *volta do mar*: the North Atlantic Gyre's clockwise circulation, south along the African coast, west across the Atlantic, home on the Westerlies. Cabral's 1500 fleet, bound for India, followed the South Atlantic Gyre's counter-clockwise currents on the same logic — and made landfall in Brazil, a current-driven event, not chance. Further south, the cold, north-flowing **Benguela Current** forced ships wide of the Angolan coast; today it drives one of the world's four major upwelling fisheries, now a documented marine-warming hotspot with direct consequences for Angola's coastal food security. Rounding the Cape, the warm **Agulhas Current**, the main channel of Indian–Atlantic water exchange, carried ships toward Mozambique; today it sustains the coral reefs, mangroves, and seagrass beds of Inhambane and the Quirimbas Archipelago. None of this is only historical: the Portuguese Current, the same gyre's Iberian branch, is a documented dominant transport pathway for microplastics along Portugal's own coast today. Full sourcing in [`field-trips/deskStudy/portugal-land-to-sea-ocean-currents-colonial-routes.md`](../../../field-trips/deskStudy/portugal-land-to-sea-ocean-currents-colonial-routes.md).

### Colonial water and land policy: ecological legacies in Brazil, Mozambique, and Angola (new in v5)

Read through the Indigenous-Science/post-colonial-justice lens above, not as neutral history. **Brazil**: coastal Atlantic Forest cleared from first contact, the Amazon interior left largely unsettled through the colonial period, then a post-independence policy arc from the 1934/1967 Forest Codes to 2004's PPCDAm monitoring programme — deforestation down to a documented 5,796 km² in the year to July 2025, the lowest in 11 years, even as fire-driven degradation rises as a share of total loss. **Mozambique**: colonial extraction reproduced almost exactly in the present by **ProSavana**, a Brazil–Japan–Mozambique agribusiness project threatening roughly four million small-scale farmers in the Nacala Corridor — constrained not by state policy but by community-led resistance from Mozambique's Justiça Ambiental and UNAC peasants' union, joined by Brazil's own MST, a direct, present-day instance of the standpoint this notebook's corrected framing centres. **Angola**: a genuinely promising late-colonial conservation start (a 1911 hunting-licence conservation fund, expanding parks and reserves through the 1930s and 1970s, ~6% of land protected by independence in 1975) collapsed entirely in the civil war that followed, the resulting vacuum filled by diamond and oil extraction continuing colonial logics under new ownership. Full sourcing in [`field-trips/deskStudy/portugal-colonial-ecological-policy-brazil-mozambique-angola.md`](../../../field-trips/deskStudy/portugal-colonial-ecological-policy-brazil-mozambique-angola.md).

### From empire to rewilding: what this means for reading the Côa Valley (new in v5)

The three subsections above are not a detour from this study's own subject. Portugal's relationship with water, at home and via the currents that carried its empire outward, has for five centuries run in one direction: extraction, first of engravings' worth of rock at Foz Côa nearly flooded for hydroelectric power, then of sugar and forest in Brazil, land and farmers' futures in Mozambique, diamonds and oil in Angola. The Greater Côa Valley rewilding effort this notebook models is, in that light, a genuinely unusual thing: Portugal choosing, on its own soil, to let a landscape's water and land relationships run the other way. That does not undo the history above, and this notebook does not claim it does. It does mean the "safe and just" framing corrected near the start of this notebook, justice and equity as the measure, not only biophysical limits, applies with equal force whether the landscape in question is the Côa Valley or the coastlines this section has just described.

Same handoff as v1: `output/rasters/` and `output/maps/` are the finished products for the StoryMap's closing chapter; `qgis/coa_eco_connectivity_ecotourism.qgz` carries every layer (now including a new "Fire history" group, and the traced tributary network in "Base context") into an interactive form.

**Known limitations, v1's list carried forward plus what's new in v2:**
- One suitability model and one parameter set per group, not Prima et al.'s full ensemble/uncertainty sweep.
- No held-out validation split; no movement-tracking data to check against.
- Air-group suitability likely reflects recording density more than true habitat preference.
- Waterway barrier severity reuses the Water resistance surface rather than a dedicated dam/weir layer.
- Land-tenure layer is descriptive context, not a validated barrier/asset classification.
- "Just" is checked against reachability and corridor impact only — no equity/accessibility data exists in this project.
- **New — catchment tracing is topology-based, not a true DEM watershed**: HydroRIVERS' generalised (`_100`) network was traced via `NEXT_DOWN` topology, not delineated from the DEM's actual flow accumulation. It's a real river network, not a fabricated one, but it isn't a HydroBASINS-style watershed polygon either.
- **New — fire data stops at ~1 July 2025**: MODIS MCD64A1 via Planetary Computer hasn't ingested later 2025/2026 fires yet (checked globally). The annual review's own 2025 fires and the 2026 Almeida-area fire are narrative-only in this notebook, not in the fire raster.
- **New — fire penalty is land-group-only**: post-fire erosion/sediment effects on the Water group are a real, plausible interaction, not modelled here.
- **New — Rewilding Portugal's 318,000ha project boundary has no polygon anywhere in either repo**: the catchment-extended study area (10,230 km², 3.2x larger) is used as a documented stand-in, not a precise match.
- **New — the beaver, Sorraia horse, Tauros, and cinereous vulture** are named as present by Rewilding Portugal's own annual review but aren't modelled as focal species in this connectivity analysis.
- **New in v3 — the Limpopo comparison is policy/literature-based, not fieldwork-based** (corrected in v5: with one specific exception — the mine-rehabilitation/water-catchment material is Linda's own first-hand professional background, see the corrected passage above and the `my-cv` skill): built from South African government, legal, and academic sources gathered this session, not first-hand site visits or data, unlike the Côa Valley material above. Its "implication for expansion" statements are options for Rewilding Portugal to evaluate, not decisions this notebook makes on the organisation's behalf.

- **New in v4 — `barrier_penalty` and `heritage_conflict_penalty` are Land/Water only, Air excluded**: stated as an explicit open item above, not a silent omission — a heritage zone's disturbance regime (tourism activity, drone/low-altitude restrictions) could plausibly matter to Air connectivity too, not modelled here.
- **New in v4 — `heritage_conflict_penalty` is flat and uniform across each polygon's full extent**: it doesn't model any gradient of actual land-use restriction within a heritage zone (e.g. a working vineyard terrace inside the Alto Douro Wine Region vs. the Côa rock-art core's tighter protection regime).
- **New in v4 — the `private_reserves` layer's Faia Brava figure (214.6 ha, OSM-sourced) is a confirmed data-quality undercount** against the published 856.87 ha figure, not yet corrected in `acquire_reserves_and_hunting_zones.py`.
- **New in v4 — Faia Brava (ATN) and Rewilding Portugal are administratively separate organisations** under a shared Rewilding Europe umbrella, not one holding: this notebook's land-tenure and safari-upscaling figures combine their footprints as one study-area total where geospatially appropriate, but any acquisition or infrastructure proposal built on this data would need separate buy-in from each organisation, flagged explicitly wherever that distinction matters above.
- **New in v4 — the safari-upscaling achievable-scale ceiling (~10% of GLTFCA) is a geometric upper bound only, not an acquisition plan**: 64.8% of the study area sits under active, non-purchasable hunting-zone licence, and the private-purchase model this project actually uses closes at a pace this notebook's own wildlife-relations comparison documents as parcel-by-parcel and generational.
- **New in v4 — the dam-cancellation counterfactual (Douro cascade vs. the 1996 Foz Côa cancellation) is stated as explicitly unresolved**: no contemporaneous source was found proving the Douro's existing capacity caused the cancellation decision.
- **New in v4 — the renewable-energy policy question (case-by-case opposition vs. a standing policy position) is left open, not resolved, by this version**: a genuine strategic tradeoff for Rewilding Portugal and the wider movement to weigh, not a decision this notebook makes on their behalf.
- **New in v4 — the integrated-infrastructure proposal (solar access canopies, photo-tourism lodges) is a desk-study sketch, not an engineering-ready design or feasibility study**: every economic figure in it is illustrative order-of-magnitude, not a real project number, and the road-canopy application is explicitly adapted from cropland agrivoltaics precedent, not an existing road-canopy precedent.
- **New in v4 — the six new comparative desk-research notes behind this version's additions are, like v3's Limpopo section, policy/literature-based, not fieldwork-based**, with two narrow exceptions: the field-photographed hunting-zone signage cross-referenced in the wildlife-relations note, and this notebook's own live-computed safari-upscaling figures above.
- **New in v6 — the off-road/track network layer's OSM tagging completeness is unverified**: `research/guarda-rewilding-access/scripts/acquire_offroad_routes.py` pulls highway=track plus surface/tracktype-tagged ways within a bounding box; areas with sparse OSM coverage will under-report track density, not because tracks don't exist there.
- **New in v6 — the Guarda-access "unpaved-fraction" ruggedness indicator is a proximity proxy, not a DEM-validated terrain-ruggedness metric**: it measures how much of each OpenRouteService driving-car route passes within 60m of an acquired unpaved track, not the literal surface of the route itself, since ORS routing avoids genuine off-road tracks by design.
